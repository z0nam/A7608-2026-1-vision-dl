```python
temp = torch.FloatTensor([1, 2, 3, 4, 5, 6, 7])
print(temp[0], temp[1], temp[-1]) # 0번째, 1번째, -1번째 원소를 출력, 파이썬은 0부터 인덱싱이 시작되고, 음수 인덱스는 뒤에서부터 원소를 선택합니다. 직관적으로 생각한다면 0이 시작 인덱스이므로 0번째 원소는 1, 1번째 원소는 2, -1번째 원소는 7이 됩니다.
print('------------------------')
print(temp[2:5], temp[4:-1]) # 2번째부터 5번째까지 (5번째는 포함하지 않음), 4번째부터 -1번째까지 (-1번째는 포함하지 않음)
print(temp[::2]) # 2칸씩 건너뛰면서 원소를 선택
print(temp[-5:-1]) # -5번째부터 -1번째까지 (마지막 원소는 포함하지 않음)
print(temp[5:2]) # 5번째부터 2번째까지 (2번째는 포함하지 않음) -> 빈 텐서가 출력됩니다. 이유는 시작 인덱스가 끝 인덱스보다 크기 때문입니다. 파이썬에서 슬라이싱은 시작 인덱스가 끝 인덱스보다 작아야 정상적으로 작동합니다. 따라서 이 경우에는 빈 텐서가 출력됩니다.
print(temp[3:]) # 3번째부터 끝까지
print(temp[:4]) # 처음부터 4번째까지 (4번째는 포함하지 않음)
print(temp[:]) # 처음부터 끝까지
print(temp[2:5:2]) # 2번째부터 5번째까지 (5번째는 포함하지 않음) 중에서 2칸씩 건너뛰면서 원소를 선택
```

```python
v = torch.tensor([1, 2, 3])
w = torch.tensor([3, 4, 6])
print(w - v) # 텐서의 뺄셈 연산을 수행합니다. 각 원소끼리 뺄셈이 이루어집니다. 따라서 결과는 [3-1, 4-2, 6-3] = [2, 2, 3]이 됩니다.
print(w * v) # 텐서의 곱셈 연산을 수행합니다. 각 원소끼리 곱셈이 이루어집니다. 따라서 결과는 [3*1, 4*2, 6*3] = [3, 8, 18]이 됩니다.
print(w / v) # 텐서의 나눗셈 연산을 수행합니다. 각 원소끼리 나눗셈이 이루어집니다. 따라서 결과는 [3/1, 4/2, 6/3] = [3.0, 2.0, 2.0]이 됩니다.
print(w ** v) # 텐서의 거듭제곱 연산을 수행합니다. 각 원소끼리 거듭제곱이 이루어집니다. 따라서 결과는 [3**1, 4**2, 6**3] = [3, 16, 216]이 됩니다.
```

```python
temp = torch.tensor([
    [1, 2], [3, 4]
])

print(temp.shape) # 텐서의 형태를 출력합니다. 이 텐서는 2행 2열로 구성되어 있으므로 (2, 2)라는 형태가 출력됩니다.
print('------------------------')
print(temp.view(4,1)) # 텐서의 형태를 변경하는 메서드입니다. 원래 텐서는 2행 2열이었지만, view(4,1)을 사용하여 4행 1열로 형태를 변경합니다. 따라서 결과는 [[1], [2], [3], [4]]가 됩니다.
print('------------------------')
print(temp.view(-1)) # 텐서를 1차원으로 평탄화합니다. 따라서 결과는 [1, 2, 3, 4]가 됩니다.
print('------------------------')
print(temp.view(1, -1)) # 텐서를 2행 2열로 형태를 변경합니다. 따라서 결과는 [[1, 2, 3, 4]]가 됩니다.   
print('------------------------')
print(temp.view(-1, 1)) # 텐서를 4행 1열로 형태를 변경합니다. 따라서 결과는 [[1], [2], [3], [4]]가 됩니다. view 메서드에서 -1은 자동으로 계산되는 차원을 나타냅니다. 따라서 view(-1, 1)은 텐서를 4행 1열로 변경하는 것을 의미합니다.
```

```python
for category in categorical_columns:
    dataset[category] = dataset[category].astype('category') # 데이터프레임의 각 열을 범주형 데이터로 변환합니다. astype('category') 메서드를 사용하여 각 열의 데이터 타입을 범주형으로 변경합니다. 범주형 데이터는 고유한 값들의 집합으로 구성된 데이터 유형으로, 머신 러닝 모델에서 효율적으로 처리할 수 있습니다. 이를 통해 모델이 각 범주를 인식하고 적절하게 처리할 수 있도록 합니다.
```

```python
class Model(nn.Module):
    def __init__(self, embedding_size, output_size, layers, p=0.4): # 모델의 초기화 메서드입니다. embedding_size는 각 범주형 열에 대한 임베딩 크기를 나타내는 리스트입니다. output_size는 모델의 출력 크기를 나타냅니다. layers는 모델의 은닉층 크기를 나타내는 리스트입니다. p는 드롭아웃 확률을 나타내며, 기본값은 0.4로 설정되어 있습니다.
        super().__init__() # 부모 클래스인 nn.Module의 초기화 메서드를 호출하여 모델을 초기화합니다. super() 함수는 부모 클래스의 메서드를 호출하는 데 사용됩니다. 따라서 super().__init__()는 nn.Module 클래스의 초기화 메서드를 호출하여 모델을 초기화하는 역할을 합니다.
        self.all_embeddings = nn.ModuleList([nn.Embedding(ni, nf) for ni, nf in embedding_size]) # embedding_size 리스트에 있는 각 열의 고유한 범주 수와 임베딩 크기를 기반으로 nn.Embedding 레이어를 생성하여 ModuleList에 저장합니다. nn.Embedding은 범주형 데이터를 임베딩 벡터로 변환하는 레이어입니다. ni는 각 열의 고유한 범주 수를 나타내며, nf는 해당 열에 대한 임베딩 크기를 나타냅니다. 따라서 self.all_embeddings 변수에는 price, maint, doors, persons, lug_capacity, safety 열 각각에 대한 nn.Embedding 레이어가 저장된 ModuleList가 됩니다.
        self.embedding_dropout = nn.Dropout(p) # 임베딩 레이어의 출력을 드롭아웃하는 레이어입니다. nn.Dropout은 주어진 확률 p에 따라 입력 텐서의 원소를 무작위로 0으로 설정하는 레이어입니다. 이를 통해 모델이 특정 입력에 과도하게 의존하는 것을 방지하고, 일반화 성능을 향상시킬 수 있습니다. 따라서 self.embedding_dropout 변수에는 드롭아웃 확률이 p로 설정된 nn.Dropout 레이어가 저장됩니다.
        
        all_layers = [] # 모델의 모든 레이어를 저장할 리스트입니다. 이 리스트에는 임베딩 레이어와 은닉층 레이어가 모두 포함됩니다. 모델의 구조를 정의하는 과정에서 각 레이어를 생성하여 all_layers 리스트에 추가하게 됩니다. 따라서 all_layers 변수는 모델의 전체 레이어 구성을 저장하는 역할을 합니다.
        num_categorical_cols = sum((nf for ni, nf in embedding_size)) # embedding_size 리스트에 있는 각 열의 임베딩 크기를 모두 더하여 num_categorical_cols 변수에 저장합니다. embedding_size는 price, maint, doors, persons, lug_capacity, safety 열 각각에 대한 고유한 범주 수와 임베딩 크기가 튜플 형태로 저장된 리스트입니다. 따라서 sum((nf for ni, nf in embedding_size))는 각 열의 임베딩 크기를 모두 더한 값을 계산하여 num_categorical_cols 변수에 저장합니다. 이 값은 모델의 입력 크기로 사용됩니다.
        input_size = num_categorical_cols 

        for i in layers: # layers 리스트에 있는 각 은닉층 크기에 대해 반복합니다. layers는 모델의 은닉층 크기를 나타내는 리스트입니다. 따라서 for i in layers:는 layers 리스트에 있는 각 은닉층 크기에 대해 반복하는 루프를 시작합니다.
            all_layers.append(nn.Linear(input_size, i)) # input_size에서 i로 가는 선형 레이어를 생성하여 all_layers 리스트에 추가합니다. nn.Linear는 선형 변환을 수행하는 레이어입니다. input_size는 이전 레이어의 출력 크기를 나타내며, i는 현재 은닉층의 크기를 나타냅니다. 따라서 nn.Linear(input_size, i)는 input_size에서 i로 가는 선형 레이어를 생성하며, 이를 all_layers 리스트에 추가합니다.
            all_layers.append(nn.ReLU(inplace=True)) # ReLU 활성화 함수를 추가합니다. nn.ReLU는 Rectified Linear Unit 활성화 함수로, 입력이 0보다 크면 그대로 출력하고, 0 이하이면 0을 출력하는 함수입니다. inplace=True는 메모리 효율성을 위해 입력 텐서를 직접 수정하여 활성화 함수를 적용하도록 하는 옵션입니다. 따라서 all_layers 리스트에는 선형 레이어 다음에 ReLU 활성화 함수가 추가됩니다.
            all_layers.append(nn.BatchNorm1d(i)) # 배치 정규화 레이어를 추가합니다. nn.BatchNorm1d는 1차원 입력에 대한 배치 정규화를 수행하는 레이어입니다. i는 현재 은닉층의 크기를 나타냅니다. 배치 정규화는 각 배치마다 입력을 정규화하여 모델의 학습을 안정화하고, 수렴 속도를 향상시키는 데 도움을 줍니다. 따라서 all_layers 리스트에는 선형 레이어 다음에 ReLU 활성화 함수와 배치 정규화 레이어가 추가됩니다.
            all_layers.append(nn.Dropout(p)) # 드롭아웃 레이어를 추가합니다. nn.Dropout는 주어진 확률 p에 따라 입력 텐서의 원소를 무작위로 0으로 설정하는 레이어입니다. 이를 통해 모델이 특정 입력에 과도하게 의존하는 것을 방지하고, 일반화 성능을 향상시킬 수 있습니다. 따라서 all_layers 리스트에는 선형 레이어 다음에 ReLU 활성화 함수, 배치 정규화 레이어, 드롭아웃 레이어가 추가됩니다.
            input_size = i # 다음 레이어의 입력 크기를 현재 은닉층의 크기로 업데이트합니다. input_size는 다음 레이어의 입력 크기를 나타내며, i는 현재 은닉층의 크기를 나타냅니다. 따라서 input_size = i는 다음 레이어의 입력 크기를 현재 은닉층의 크기로 업데이트하는 역할을 합니다. 이를 통해 모델의 레이어가 순차적으로 연결될 수 있도록 합니다.

        all_layers.append(nn.Linear(layers[-1], output_size)) # 마지막 은닉층에서 output_size로 가는 선형 레이어를 생성하여 all_layers 리스트에 추가합니다. nn.Linear는 선형 변환을 수행하는 레이어입니다. layers[-1]은 layers 리스트의 마지막 요소로, 마지막 은닉층의 크기를 나타냅니다. output_size는 모델의 출력 크기를 나타냅니다. 따라서 nn.Linear(layers[-1], output_size)는 마지막 은닉층에서 output_size로 가는 선형 레이어를 생성하며, 이를 all_layers 리스트에 추가합니다.
        self.layers = nn.Sequential(*all_layers) # all_layers 리스트에 저장된 레이어들을 nn.Sequential로 묶어서 self.layers 변수에 저장합니다. nn.Sequential은 주어진 레이어들을 순차적으로 연결하는 컨테이너입니다. *all_layers는 리스트에 저장된 레이어들을 언패킹하여 nn.Sequential에 전달하는 역할을 합니다. 따라서 self.layers 변수에는 all_layers 리스트에 저장된 레이어들이 순차적으로 연결된 형태의 nn.Sequential 객체가 저장됩니다.

    def forward(self, x_categorical): # 모델의 순전파 메서드입니다. x_categorical은 범주형 입력 데이터를 나타내는 텐서입니다. 이 메서드는 모델에 입력된 범주형 데이터를 처리하여 최종 출력값을 반환하는 역할을 합니다.
        embeddings = [] # 임베딩된 벡터를 저장할 리스트입니다. 이 리스트에는 각 범주형 열에 대한 임베딩 벡터가 저장됩니다. 모델의 순전파 과정에서 각 범주형 열에 대해 nn.Embedding 레이어를 사용하여 임베딩 벡터를 생성하고, 이를 embeddings 리스트에 추가하게 됩니다. 따라서 embeddings 변수는 모델의 입력으로 사용되는 범주형 데이터가 임베딩된 형태로 저장되는 리스트입니다.
        for i,e in enumerate(self.all_embeddings): # self.all_embeddings에 저장된 각 nn.Embedding 레이어에 대해 반복합니다. enumerate() 함수를 사용하여 self.all_embeddings 리스트의 각 요소와 해당 요소의 인덱스를 동시에 가져옵니다. i는 현재 레이어의 인덱스를 나타내며, e는 현재 레이어인 nn.Embedding 객체를 나타냅니다. 따라서 for i,e in enumerate(self.all_embeddings):는 self.all_embeddings 리스트에 저장된 각 nn.Embedding 레이어에 대해 반복하는 루프를 시작합니다.
            embeddings.append(e(x_categorical[:,i])) # x_categorical 텐서에서 i번째 열을 선택하여 nn.Embedding 레이어 e에 입력으로 전달하고, 생성된 임베딩 벡터를 embeddings 리스트에 추가합니다. x_categorical[:,i]는 x_categorical 텐서에서 모든 행에 대해 i번째 열을 선택하는 슬라이싱입니다. e(x_categorical[:,i])는 nn.Embedding 레이어 e에 선택된 열을 입력으로 전달하여 해당 열에 대한 임베딩 벡터를 생성하는 역할을 합니다. 따라서 embeddings.append(e(x_categorical[:,i]))는 각 범주형 열에 대한 임베딩 벡터를 생성하여 embeddings 리스트에 추가하는 역할을 합니다.
        x = torch.cat(embeddings, 1) # embeddings 리스트에 저장된 임베딩 벡터들을 세로 방향으로 이어붙입니다. torch.cat은 주어진 텐서들을 지정된 축을 따라 이어붙이는 함수입니다. 여기서는 axis=1로 지정하여 각 임베딩 벡터가 열로 이어붙여지도록 합니다. 따라서 x 변수에는 각 범주형 열에 대한 임베딩 벡터들이 세로 방향으로 이어붙여진 형태의 텐서가 저장됩니다.
        x = self.embedding_dropout(x) # 임베딩 벡터에 드롭아웃을 적용합니다. self.embedding_dropout는 nn.Dropout 레이어로, 주어진 확률 p에 따라 입력 텐서의 원소를 무작위로 0으로 설정하는 역할을 합니다. 이를 통해 모델이 특정 입력에 과도하게 의존하는 것을 방지하고, 일반화 성능을 향상시킬 수 있습니다. 따라서 x = self.embedding_dropout(x)는 임베딩 벡터에 드롭아웃을 적용하여 모델의 학습을 안정화하는 역할을 합니다.
        x = self.layers(x) # self.layers에 저장된 nn.Sequential 객체를 사용하여 x 텐서를 순차적으로 처리합니다. self.layers는 all_layers 리스트에 저장된 레이어들이 순차적으로 연결된 형태의 nn.Sequential 객체입니다. 따라서 x = self.layers(x)는 x 텐서를 self.layers에 저장된 레이어들을 순차적으로 통과시키면서 최종 출력값을 계산하는 역할을 합니다.
        return x # 최종 출력값을 반환합니다. 모델의 순전파 과정에서 입력된 범주형 데이터가 임베딩된 형태로 처리되고, self.layers에 저장된 레이어들을 통과하여 최종 출력값이 계산됩니다. 따라서 return x는 모델의 최종 출력값을 반환하는 역할을 합니다.
```

```python
model = Model(categorical_embedding_sizes, 4, [200,100,50], p=0.4) # Model 클래스의 인스턴스를 생성합니다. categorical_embedding_sizes는 각 범주형 열에 대한 고유한 범주 수와 임베딩 크기가 튜플 형태로 저장된 리스트입니다. 4는 모델의 출력 크기를 나타냅니다. [200, 100, 50]은 모델의 은닉층 크기를 나타내는 리스트입니다. p=0.4는 드롭아웃 확률을 나타내며, 기본값은 0.4로 설정되어 있습니다. 따라서 model 변수에는 Model 클래스의 인스턴스가 저장됩니다.
print(model) # model 변수에 저장된 Model 클래스의 인스턴스의 구조를 출력합니다. Model 클래스는 nn.Module을 상속받아 정의된 모델로, __init__ 메서드에서 모델의 레이어 구성을 정의하고, forward 메서드에서 순전파 과정을 구현합니다. 따라서 print(model)은 모델의 레이어 구성과 순전파 과정을 포함한 모델의 구조를 출력하는 역할을 합니다.
```

```python
loss_function = nn.CrossEntropyLoss() # 모델의 손실 함수를 정의합니다. nn.CrossEntropyLoss는 다중 클래스 분류 문제에서 자주 사용되는 손실 함수로, 모델의 출력과 실제 레이블 간의 차이를 계산하여 모델을 학습시키는 데 사용됩니다. 따라서 loss_function 변수에는 nn.CrossEntropyLoss 객체가 저장됩니다.
optimizer = torch.optim.Adam(model.parameters(), lr=0.001) # 모델의 최적화 알고리즘을 정의합니다. torch.optim.Adam은 Adam 최적화 알고리즘을 구현한 클래스입니다. model.parameters()는 모델의 학습 가능한 매개변수를 반환하는 메서드입니다. lr=0.001은 학습률을 나타내며, 모델이 가중치를 업데이트하는 속도를 제어합니다. 따라서 optimizer 변수에는 Adam 최적화 알고리즘이 적용된 옵티마이저 객체가 저장됩니다.
epochs = 200 # 모델을 학습시킬 에포크 수를 정의합니다. 에포크는 전체 데이터셋을 한 번 학습시키는 단위를 나타냅니다. 따라서 epochs 변수에는 모델을 200번 학습시키겠다는 의미입니다.
```

```python
if torch.cuda.is_available():
    device = torch.device('cuda')
else:
    device = torch.device('cpu')

print("device=", device)
```

```python
epochs = 500
aggregated_losses = [] # 모델의 학습 과정에서 발생하는 손실 값을 저장할 리스트입니다. 각 에포크마다 계산된 손실 값을 이 리스트에 추가하여, 학습 과정에서 손실이 어떻게 변화하는지 추적할 수 있습니다. 따라서 aggregated_losses 변수에는 모델의 학습 과정에서 발생하는 손실 값이 저장된 리스트가 됩니다.
train_outputs = train_outputs.to(device=device, dtype=torch.int64) # train_outputs 텐서를 지정된 장치로 이동시키고, 데이터 타입을 64비트 정수형으로 변환합니다. to() 메서드는 텐서를 지정된 장치로 이동시키는 역할을 합니다. device는 모델이 학습되는 장치를 나타내며, dtype=torch.int64는 텐서의 데이터 타입을 64비트 정수형으로 지정하는 옵션입니다. 따라서 train_outputs 변수에는 train_outputs 텐서가 지정된 장치로 이동되고, 데이터 타입이 64비트 정수형으로 변환된 형태가 저장됩니다.
for i in range(epochs):
    i += 1
    y_pred = model(categorical_train_data).to(device) # categorical_train_data 텐서를 모델에 입력으로 전달하여 예측값을 계산하고, 이를 지정된 장치로 이동시킵니다. model(categorical_train_data)는 모델의 순전파 메서드를 호출하여 categorical_train_data 텐서를 처리하고, 예측값을 계산하는 역할을 합니다. to(device) 메서드는 계산된 예측값을 지정된 장치로 이동시키는 역할을 합니다. 따라서 y_pred 변수에는 모델이 categorical_train_data를 입력으로 처리하여 계산한 예측값이 지정된 장치로 이동된 형태가 저장됩니다.
    single_loss = loss_function(y_pred, train_outputs) # 모델의 예측값 y_pred와 실제 레이블 train_outputs 간의 손실 값을 계산합니다. loss_function은 nn.CrossEntropyLoss 객체로, 다중 클래스 분류 문제에서 자주 사용되는 손실 함수입니다. y_pred는 모델이 categorical_train_data를 입력으로 처리하여 계산한 예측값을 나타내며, train_outputs는 실제 레이블을 나타냅니다. 따라서 single_loss 변수에는 모델의 예측값과 실제 레이블 간의 손실 값이 저장됩니다.
    aggregated_losses.append(single_loss)

    if i%25 == 1:
        print(f'epoch: {i:3} loss: {single_loss.item():10.8f}') # 25 에포크마다 현재 에포크 번호와 손실 값을 출력합니다. i%25 == 1은 i가 25로 나누어 떨어지는 경우에만 출력하도록 하는 조건입니다. single_loss.item()은 single_loss 텐서에서 스칼라 값을 추출하는 메서드입니다. 따라서 print(f'epoch: {i:3} loss: {single_loss.item():10.8f}')는 현재 에포크 번호와 손실 값을 형식화하여 출력하는 역할을 합니다.

    optimizer.zero_grad() # 옵티마이저의 기울기를 초기화합니다. optimizer.zero_grad()는 모델의 매개변수에 저장된 기울기를 0으로 초기화하는 역할을 합니다. 이는 모델을 학습시키는 과정에서 이전 에포크에서 계산된 기울기가 다음 에포크에 영향을 미치지 않도록 하기 위해 필요합니다. 따라서 optimizer.zero_grad()를 호출하여 모델의 매개변수에 저장된 기울기를 초기화합니다.
    single_loss.backward() # 손실 값에 대한 모델의 매개변수의 기울기를 계산합니다. single_loss.backward()는 손실 값인 single_loss에 대해 모델의 매개변수에 대한 기울기를 계산하는 역할을 합니다. 이 메서드는 자동 미분 기능을 사용하여 손실 값이 모델의 매개변수에 어떻게 영향을 미치는지 계산합니다. 따라서 single_loss.backward()를 호출하여 모델의 매개변수에 대한 기울기를 계산합니다.
    optimizer.step() # 모델의 매개변수를 업데이트합니다. optimizer.step()는 계산된 기울기를 사용하여 모델의 매개변수를 업데이트하는 역할을 합니다. 이 메서드는 옵티마이저가 정의된 최적화 알고리즘에 따라 매개변수를 조정하여 손실 값을 최소화하려고 시도합니다. 따라서 optimizer.step()을 호출하여 모델의 매개변수를 업데이트합니다.

print(f'epoch: {i:3} loss: {single_loss.item():10.10f}') # 마지막 에포크의 번호와 손실 값을 출력합니다. i는 마지막 에포크 번호를 나타내며, single_loss.item()은 single_loss 텐서에서 스칼라 값을 추출하는 메서드입니다. 따라서 print(f'epoch: {i:3} loss: {single_loss.item():10.10f}')는 마지막 에포크 번호와 손실 값을 형식화하여 출력하는 역할을 합니다.
```

```python
test_outputs = test_outputs.to(device=device, dtype=torch.int64) # test_outputs 텐서를 지정된 장치로 이동시키고, 데이터 타입을 64비트 정수형으로 변환합니다. to() 메서드는 텐서를 지정된 장치로 이동시키는 역할을 합니다. device는 모델이 학습되는 장치를 나타내며, dtype=torch.int64는 텐서의 데이터 타입을 64비트 정수형으로 지정하는 옵션입니다. 따라서 test_outputs 변수에는 test_outputs 텐서가 지정된 장치로 이동되고, 데이터 타입이 64비트 정수형으로 변환된 형태가 저장됩니다.
with torch.no_grad(): # torch.no_grad()는 해당 블록 내에서 자동 미분 기능을 비활성화하는 컨텍스트 매니저입니다. 이는 모델의 예측을 계산할 때 기울기를 계산하지 않도록 하여 메모리 사용량을 줄이고, 계산 속도를 향상시키는 데 도움을 줍니다. 따라서 with torch.no_grad(): 블록 내에서는 모델의 예측을 계산할 때 기울기가 계산되지 않도록 합니다.
    y_val = model(categorical_test_data).to(device) # categorical_test_data 텐서를 모델에 입력으로 전달하여 예측값을 계산하고, 이를 지정된 장치로 이동시킵니다. torch.no_grad()는 해당 블록 내에서 자동 미분 기능을 비활성화하는 컨텍스트 매니저입니다. 이는 모델의 예측을 계산할 때 기울기를 계산하지 않도록 하여 메모리 사용량을 줄이고, 계산 속도를 향상시키는 데 도움을 줍니다. model(categorical_test_data)는 모델의 순전파 메서드를 호출하여 categorical_test_data 텐서를 처리하고, 예측값을 계산하는 역할을 합니다. to(device) 메서드는 계산된 예측값을 지정된 장치로 이동시키는 역할을 합니다. 따라서 y_val 변수에는 모델이 categorical_test_data를 입력으로 처리하여 계산한 예측값이 지정된 장치로 이동된 형태가 저장됩니다.
    loss = loss_function(y_val, test_outputs) # 모델의 예측값 y_val과 실제 레이블 test_outputs 간의 손실 값을 계산합니다. loss_function은 nn.CrossEntropyLoss 객체로, 다중 클래스 분류 문제에서 자주 사용되는 손실 함수입니다. y_val은 모델이 categorical_test_data를 입력으로 처리하여 계산한 예측값을 나타내며, test_outputs는 실제 레이블을 나타냅니다. 따라서 loss 변수에는 모델의 예측값과 실제 레이블 간의 손실 값이 저장됩니다.
print(f'Loss: {loss:.8f}')
```
